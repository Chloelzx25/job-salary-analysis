## Drop Junk Columns

We keep only `job_id` from postings1 since it contains junk columns (Easter Egg and Unimportant Column).
We use postings5 as the clean source for `company_name` and `title` instead.

```python
df1 = df1[['job_id']]
df5 = df5[['job_id', 'company_name', 'title']]
```

##  Merge All Files

We merge all 5 files into one dataframe using `job_id` as the common key.
A left join ensures no job listings are lost even if some files have missing records.

```python
df = df1.merge(df5, on='job_id', how='left')
df = df.merge(df2, on='job_id', how='left')
df = df.merge(df3, on='job_id', how='left')
df = df.merge(df4, on='job_id', how='left')
print(df.shape)
```

```python
df
```

## Normalise Salary to Annual Figures

Not all salaries are in the same pay period — some are hourly, monthly etc.
We convert all to annual so comparisons are fair:
- HOURLY × 2080 (40hrs × 52 weeks)
- MONTHLY × 12
- WEEKLY × 52
- BIWEEKLY × 26

```python
CONVERSION = {
    'YEARLY': 1,
    'HOURLY': 2080,
    'MONTHLY': 12,
    'WEEKLY': 52,
    'BIWEEKLY': 26
}

df['annual_salary'] = df.apply(
    lambda row: row['min_salary'] * CONVERSION.get(row['pay_period'], 1)
    if pd.notnull(row['min_salary']) and pd.notnull(row['pay_period']) else None,
    axis=1
)
```

## Remove Nulls and Outliers

We drop rows where our 3 key research variables are missing.
Without salary, experience level, or work type we cannot answer the research question.
Salary is capped between `$15,000` and `$500,000` to remove clearly incorrect data entries.

```python
df = df.drop(columns=[
    'description', 'skills_desc', 'posting_domain', 'sponsored',
    'work_type', 'currency', 'compensation_type', 'closed_time',
    'expiry', 'company_id', 'job_posting_url', 'application_url',
    'application_type', 'original_listed_time', 'remote_allowed',
    'views', 'applies', 'listed_time'
], errors='ignore')
```

```python
df_clean = df.dropna(subset=['annual_salary', 'formatted_experience_level', 'formatted_work_type'])
df_clean = df_clean[df_clean['annual_salary'] <= 500000]
df_clean = df_clean[df_clean['annual_salary'] >= 15000]

print(f'Rows ready for analysis: {len(df_clean)}')
print(df_clean['annual_salary'].describe())
```

## Self-Researched Matplotlib Feature — ax.axvline()

`ax.axvline()` draws a vertical line at a specified x-axis position. It was not 
taught in class. I used it in the histogram to mark the mean and median salary, 
making it immediately clear whether the distribution is skewed and where most 
salaries cluster.

## Chart 1 — Histogram: Distribution of Annual Salaries

Before comparing salary by experience level, work type, and location, we first
need to understand the overall shape of salary data — where most salaries
cluster and whether the distribution is skewed.

The mean and median lines use ax.axvline() (self-researched Matplotlib feature)
to show that the distribution is right-skewed. This means a small number of
very high-paying senior roles pull the mean above the median, and most job
seekers can expect a salary closer to the median value.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))

ax.hist(df_clean['annual_salary'], bins=40, color='steelblue', edgecolor='white')

# Self-researched: ax.axvline() draws a vertical line at a given x position
ax.axvline(df_clean['annual_salary'].mean(), color='red', linestyle='--',
           label=f"Mean: ${df_clean['annual_salary'].mean():,.0f}")
ax.axvline(df_clean['annual_salary'].median(), color='orange', linestyle='--',
           label=f"Median: ${df_clean['annual_salary'].median():,.0f}")

ax.set_title('Distribution of Annual Salaries', fontsize=14)
ax.set_xlabel('Annual Salary (USD)')
ax.set_ylabel('Count')
ax.legend()
plt.tight_layout()
plt.savefig('chart1_histogram.png', dpi=150)
plt.show()
```

## Self-Researched Pandas Method — groupby().agg()

`groupby()` groups rows by a specified column and `.agg()` applies multiple 
aggregation functions at once. It was not taught in class.

I used it to compute the average salary grouped by experience level and work 
type, which directly answers the research question by showing how salary varies 
across these variables.

Without groupby(), we would have to manually filter and calculate each group 
separately — groupby() does this in one step.

## Chart 2 — Bar Chart: Average Salary by Location (Top 10)

This bar chart directly addresses the location variable in our research question
by showing which locations offer the highest average annual salary.

By identifying the top paying locations, job seekers can make informed decisions
about where to target their job search for maximum earning potential.

```python
top_locations = (
    df_clean.groupby('location')['annual_salary']
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .round(0)
)

fig, ax = plt.subplots(figsize=(12, 5))
bars = ax.bar(top_locations.index, top_locations.values, color='steelblue', edgecolor='white')

# Add value labels on top of each bar
for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 500,
            f'${h:,.0f}', ha='center', fontsize=8)

ax.set_title('Top 10 Locations by Average Annual Salary', fontsize=14)
ax.set_xlabel('Location')
ax.set_ylabel('Average Annual Salary (USD)')
ax.set_xticks(range(len(top_locations.index)))
ax.set_xticklabels(top_locations.index, rotation=30, ha='right')
plt.tight_layout()
plt.savefig('chart2_bar.png', dpi=150)
plt.show()
```

## Chart 3 — Pie Chart: Proportion of Average Salary by Experience Level

This pie chart addresses the experience level variable in our research question
by showing what proportion of the total average salary each experience level
represents.

Higher experience levels take up a larger slice of the pie, confirming that
salary grows significantly with seniority — directly answering how salary
varies by experience level.

```python
exp_salary = (
    df_clean.groupby('formatted_experience_level')['annual_salary']
    .mean()
    .round(0)
)

fig, ax = plt.subplots(figsize=(9, 7))
wedges, texts, autotexts = ax.pie(
    exp_salary.values,
    autopct=lambda pct: f'{pct:.1f}%' if pct > 3 else '',
    colors=['steelblue', '#6ACC65', '#D65F5F', '#B47CC7', '#C4AD66', '#77BEDB'],
    startangle=140,
    pctdistance=0.75
)

ax.legend(wedges, exp_salary.index, title='Experience Level',
          loc='center left', bbox_to_anchor=(1, 0, 0.5, 1))
ax.set_title('Proportion of Average Salary by Experience Level', fontsize=14)
plt.tight_layout()
plt.savefig('chart3_pie.png', dpi=150)
plt.show()
```

## Chart 4 — Boxplot: Salary Distribution by Work Type

This boxplot directly addresses the work type variable in our research question.
Unlike a bar chart which only shows averages, a boxplot shows the median,
spread, and outliers for each work type.

This helps us see not just which work type pays more on average, but how
consistent or variable the salaries are — important for job seekers weighing
stability against earning potential.

```python
main_types = ['Full-time', 'Contract', 'Part-time', 'Internship']
boxplot_data = [
    df_clean[df_clean['formatted_work_type'] == wt]['annual_salary'].dropna().values
    for wt in main_types
]

fig, ax = plt.subplots(figsize=(10, 5))
bp = ax.boxplot(
    boxplot_data,
    tick_labels=main_types,
    patch_artist=True,
    medianprops=dict(color='black', linewidth=2),
    flierprops=dict(marker='o', markersize=3, alpha=0.6,
                    markeredgewidth=0, markerfacecolor='gray')
)

colours = ['steelblue', '#6ACC65', '#D65F5F', '#B47CC7']
for patch, colour in zip(bp['boxes'], colours):
    patch.set_facecolor(colour)
    patch.set_alpha(0.7)

ax.set_ylim(0, 300000)
ax.set_title('Salary Distribution by Work Type', fontsize=14)
ax.set_xlabel('Work Type')
ax.set_ylabel('Annual Salary (USD)')
plt.tight_layout()
plt.savefig('chart4_boxplot.png', dpi=150)
plt.show()
```

## Chart 5 — Line Chart: Average Salary by Experience Level and Work Type

This line chart is the strongest chart for our research question as it combines
experience level and work type together in one visualisation — showing how
salary grows across career stages for each work type.

Combined with Chart 2 (location), all three research variables are fully
addressed across the 5 charts, giving a complete answer to our research question.

```python
exp_order = ['Internship', 'Entry level', 'Associate', 'Mid-Senior level', 'Director', 'Executive']
main_types = ['Full-time', 'Contract', 'Part-time']
colours = ['steelblue', '#6ACC65', '#D65F5F']

fig, ax = plt.subplots(figsize=(11, 5))

for wt, colour in zip(main_types, colours):
    subset = df_clean[df_clean['formatted_work_type'] == wt]
    # Self-researched: groupby().agg() computes aggregations in one step
    avg = subset.groupby('formatted_experience_level')['annual_salary'].agg('mean').reindex(exp_order)
    ax.plot(exp_order, avg.values, marker='o', label=wt, color=colour, linewidth=2)

ax.set_title('Average Salary by Experience Level and Work Type', fontsize=14)
ax.set_xlabel('Experience Level')
ax.set_ylabel('Average Annual Salary (USD)')
ax.set_xticks(range(len(exp_order)))
ax.set_xticklabels(exp_order, rotation=15, ha='right')
ax.legend(title='Work Type')
plt.tight_layout()
plt.savefig('chart5_line.png', dpi=150)
plt.show()
```
